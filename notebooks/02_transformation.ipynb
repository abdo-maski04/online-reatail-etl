```python
import pandas as pd
```

```python
db=pd.read_excel("../data/raw/Online Retail.xlsx")
```

```python
sales=db.copy()
```

```python
print(sales)
```

```python
# on supprimes les lignes avec des valeurs manquantes soit pour description soit customerId
sales=sales.dropna()
# calcules le total des lignes dupliqués
nb=sales.isnull().sum()
print(nb)
```

```python
# supprimant les doublons
nbr_dbl=sales.duplicated().sum()
print(nbr_dbl)
sales=sales.drop_duplicates(keep='first')
nbr_dbl=sales.duplicated().sum()
print(nbr_dbl)
```

```python
# on ajout un colonne total_price=Quantty*UnitPrice
sales["totalPrice"]=sales["Quantity"]*sales["UnitPrice"]
```

```python
# on supprime les lignes ayant un quantity négatif et prix unitaire == 0
sales=sales[(sales["Quantity"]>0) & (sales["UnitPrice"]!=0)]
```

```python
sales.info()
# on corrige les types
sales["InvoiceNo"]=sales["InvoiceNo"].astype(str)
sales["Description"]=sales["Description"].astype(str)
sales["StockCode"]=sales["StockCode"].astype(str)
sales["CustomerID"]=sales["CustomerID"].astype(int)
```

```python
sales.info()
```

```python
# on définie de DataFrame une continne les canceled sales et autre les sales
#on a supprimé les lignes des commandes annulés car chaqune est annulé leur quantité est négative mais on peux stocker ces revenue dans une DataFrame pour les analyser et les comparer avec les ventes réelles
# on utilise db original DataFrame
revenues=db.copy()
revenues=revenues[revenues["InvoiceNo"].str.startswith(("C"),na=False)]
print(revenues)
```

```python
# on supprimes les lignes avec des valeurs manquantes soit pour description soit customerId depuis revenues
revenues=revenues.dropna()
# calcules le total des lignes dupliqués
nb=revenues.isnull().sum()
print(nb)
```

```python
# supprimant les doublons
nbr_dbl=revenues.duplicated().sum()
print(nbr_dbl)
revenues=revenues.drop_duplicates(keep='first')
nbr_dbl=revenues.duplicated().sum()
print(nbr_dbl)
```

```python
# on supprime les lignes avec un prix unitaire == 0 depuis revenues
revenues=revenues[revenues["UnitPrice"]!=0]
```

```python
# on ajout un colonne total_price=Quantty*UnitPrice dans revenues
revenues["totalPrice"]=revenues["Quantity"]*revenues["UnitPrice"]
```

```python
# on corrige les types dans revenues
revenues["InvoiceNo"]=revenues["InvoiceNo"].astype(str)
revenues["Description"]=revenues["Description"].astype(str)
revenues["StockCode"]=revenues["StockCode"].astype(str)
revenues["CustomerID"]=revenues["CustomerID"].astype(int)
```

```python
sales.to_csv("../data/clean/clean_data_sales.csv",index=False)
revenues.to_csv("../data/clean/clean_data_revenues.csv",index=False)
```
